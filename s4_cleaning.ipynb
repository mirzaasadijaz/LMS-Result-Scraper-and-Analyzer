```python
courses = [col.replace('Marks_', '') for col in df.columns if col.startswith('Marks_')]

new_order = ['Registration Number', 'Name']

for course in courses:
    new_order.extend([f'Marks_{course}', f'Grade_{course}'])

df = df[new_order]
```

```python
credit_hours = {
    'BMS-402': 2,
    'CS-406': 3,
    'CS-408': 3,
    'CS-410': 3,
    'CS-412': 3,
    'IS-402': 0,
    'STAT-402': 3
}
total_credits = sum(credit_hours.values())

# For 1 Credit Hour Courses (Total 20 Marks)
qp_1_credit = {
    8: 1.00, 9: 1.50, 10: 2.00, 11: 2.33, 12: 2.67, 
    13: 3.00, 14: 3.33, 15: 3.67, 
    16: 4.00, 17: 4.00, 18: 4.00, 19: 4.00, 20: 4.00
}

# For 2 Credit Hour Courses (Total 40 Marks)
qp_2_credit = {
    16: 2.00, 17: 2.50, 18: 3.00, 19: 3.50, 20: 4.00,
    21: 4.33, 22: 4.67, 23: 5.00, 24: 5.33, 25: 5.67,
    26: 6.00, 27: 6.33, 28: 6.67, 29: 7.00, 30: 7.33, 31: 7.67, 
    32: 8.00, 33: 8.00, 34: 8.00, 35: 8.00, 36: 8.00, 
    37: 8.00, 38: 8.00, 39: 8.00, 40: 8.00
}

# For 3 Credit Hour Courses (Total 60 Marks)
qp_3_credit = {
    24: 3.00, 25: 3.50, 26: 4.00, 27: 4.50, 28: 5.00,
    29: 5.50, 30: 6.00, 31: 6.33, 32: 6.67, 33: 7.00,
    34: 7.33, 35: 7.67, 36: 8.00, 37: 8.33, 38: 8.67,
    39: 9.00, 40: 9.33, 41: 9.67, 42: 10.00, 43: 10.33,
    44: 10.67, 45: 11.00, 46: 11.33, 47: 11.67, 
    48: 12.00, 49: 12.00, 50: 12.00, 51: 12.00, 52: 12.00, 
    53: 12.00, 54: 12.00, 55: 12.00, 56: 12.00, 57: 12.00, 
    58: 12.00, 59: 12.00, 60: 12.00
}

# For 4 Credit Hour Courses (Total 80 Marks)
qp_4_credit = {
    32: 4.00, 33: 4.50, 34: 5.00, 35: 5.50, 36: 6.00,
    37: 6.50, 38: 7.00, 39: 7.50, 40: 8.00, 41: 8.33,
    42: 8.67, 43: 9.00, 44: 9.33, 45: 9.67, 46: 10.00,
    47: 10.33, 48: 10.67, 49: 11.00, 50: 11.33, 51: 11.67,
    52: 12.00, 53: 12.33, 54: 12.67, 55: 13.00, 56: 13.33,
    57: 13.67, 58: 14.00, 59: 14.33, 60: 14.67, 61: 15.00,
    62: 15.33, 63: 15.67, 64: 16.00, 65: 16.00, 66: 16.00, 
    67: 16.00, 68: 16.00, 69: 16.00, 70: 16.00, 71: 16.00,
    72: 16.00, 73: 16.00, 74: 16.00, 75: 16.00, 76: 16.00,
    77: 16.00, 78: 16.00, 79: 16.00, 80: 16.00
}

# 1. Group your dictionaries
qp_dicts = {
    1: qp_1_credit,
    2: qp_2_credit,
    3: qp_3_credit,
    4: qp_4_credit
}

# 1. Initialize a temporary running total with zeros for every student
running_total_qp = pd.Series(0.0, index=df.index)

# 2. Loop through every course
for course, credits in credit_hours.items():
    marks_col = f'Marks_{course}'
    grade_col = f'Grade_{course}'  # Now we look at the Grade column too!
    
    # Check if both columns actually exist in the dataframe to prevent errors
    if marks_col in df.columns and grade_col in df.columns:
        if credits > 0:
            # Calculate the QP for this course based on the marks
            course_qp = df[marks_col].map(qp_dicts[credits]).fillna(0.00)
            
            # --- THE NEW LOGIC ---
            # If the student's grade is 'F' or 'P', force their points to exactly 0.00
            course_qp = course_qp.mask(df[grade_col].isin(['F', 'P']), 0.00)
            
            # Add these checked points to the running total
            running_total_qp += course_qp
            
        # (If credits == 0 like IS-402, it adds 0 points automatically by skipping)

# 3. Assign the final accumulated points directly to the dataframe
df['Total_QP'] = running_total_qp

# 4. Calculate Final GPA
df['GPA'] = (df['Total_QP'] / total_credits).round(2)

df[['Registration Number', 'Name', 'Total_QP', 'GPA']].head()
```
